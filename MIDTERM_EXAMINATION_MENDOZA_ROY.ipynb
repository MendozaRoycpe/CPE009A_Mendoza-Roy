Write a program that uses nested loops to print this pattern:

*# 1

**# 2

***# 3

****# 4

*****# 5

****# 4

***# 3

**# 2

*# 1

```python
#list of the variable and its content
scores = (1,2,3,4,5)
ast = ("*")

# Code
for x in scores:
  for y in ast:
    print((y * x) + str(x))
    if x == 5:
      break
for x in range(4,0,-1):
  for y in ast:
    print((y * x) + str(x))
```
