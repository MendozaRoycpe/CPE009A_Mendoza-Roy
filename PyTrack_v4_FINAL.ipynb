 Student Info Class Summary

This class, `StudentInfo`, acts as a blueprint for creating student objects. Each object holds several key pieces of information about a student:

*   **`name`**: The student's full name.
*   **`gender`**: The student's gender (e.g., 'M' or 'F').
*   **`id`**: A unique identification number for the student.
*   **`quiz`**: The student's quiz grade.
*   **`recitation`**: The student's recitation grade.
*   **`exam`**: The student's exam grade.
*   **`finalGrade`**: The calculated final grade based on weighted averages of quiz, recitation, and exam scores.
*   **`occupancy`**: A boolean flag indicating whether this student slot in the `students` list is currently occupied by an active student record (`True`) or is available (`False`).

The `__init__` method initializes these attributes with default values (empty strings, 0, or 0.0) when a new `StudentInfo` object is created.

### Global Variables Summary

This section initializes global variables that are used throughout the program:

*   **`studentSize`**: An integer defining the maximum capacity of student records the system can manage. It is set to `100`.
*   **`students`**: A list that serves as the primary storage for all student records. It is initialized with `studentSize` (100) `StudentInfo` objects. Each of these initial objects has its `occupancy` attribute set to `False`, indicating that no student data is stored there yet.
*   **`counter`**: An integer variable that keeps track of the number of currently active (occupied) student records in the `students` list. It is initialized to `0` because no students are added at the beginning of the program's execution.

```python
import os

class StudentInfo:

    def __init__(self):
        self.name = ""
        self.gender = ""
        self.id = 0
        self.quiz = 0.0
        self.recitation = 0.0
        self.exam = 0.0
        self.finalGrade = 0.0
        self.occupancy = False

studentSize = 100
students = [StudentInfo() for _ in range(studentSize)]

counter = 0
```

### `clear()` Function

This function is used to clear the console screen. It detects the operating system (`'nt'` for Windows, otherwise assumed to be Unix-like) and executes the appropriate command (`'cls'` or `'clear'`) to clear the terminal output. This provides a cleaner user interface during program interaction.

```python
def clear():
    os.system('cls' if os.name == 'nt' else 'clear')
```

### `confirm_enter()` Function Summary

This function prompts the user to confirm an action by asking 'Are you sure you want to enter? (Yes/No):'. It enters a loop to ensure the user provides a valid input ('Yes' or 'No'). It returns `True` if the user confirms 'Yes' and `False` if the user confirms 'No'. Invalid inputs are met with a prompt to enter only 'Yes' or 'No'.

```python
def confirm_enter():
    while True:
        choice = input("Are you sure you want to enter? (Yes/No): ").strip().lower()

        if choice == "yes":
            return True
        elif choice == "no":
            return False
        else:
            print("Please enter ONLY 'Yes' or 'No'.")
```

### `userType()` Function Summary

This function is the main entry point for user interaction, allowing users to log in as an Instructor, Student, or to Quit the program. It repeatedly prompts the user for input until 'Q' is chosen:

*   If 'I' (Instructor) is entered, it calls `confirm_enter()` for verification before proceeding to `professorMenu()`.
*   If 'S' (Student) is entered, it takes a student ID, verifies if the ID exists in the `students` list, and then calls `studentMenu()` with the verified ID.
*   If 'Q' (Quit) is entered, it breaks the loop and ends the program with a thank you message.
*   For any other input, it displays an 'Invalid Input' message.

```python
def userType():
    while True:
        user = input("Log in as Instructor or Student or Quit? (I/S/Q): ").upper()

        if user == 'I':
            if confirm_enter():
                professorMenu()
            else:
                print("Cancelled.\n")

        elif user == 'S':
            entry = studentUser()
            verify = any(s.id == entry for s in students)

            if verify:
                studentMenu(entry)
            else:
                print("Student ID not found.")

        elif user == 'Q':
            print("Thank you for using the program.")
            break

        else:
            print("Invalid Input.")
```

### `professorMenu()` Function Summary

This function provides the menu of actions available to an instructor. It displays options such as adding, editing, deleting, or viewing student information, and logging out. Based on the instructor's numerical choice, it calls the corresponding function (`add()`, `edit()`, `delete()`, `view()`). Choosing '5' logs out the instructor from this menu. Invalid choices are handled with an 'Invalid Action' message.

```python
def professorMenu():
    while True:
        print("\nProfessor Menu")
        print("(1) - Add student")
        print("(2) - Edit student's Grade")
        print("(3) - Delete student Info")
        print("(4) - View student")
        print("(5) - Log-out")

        choice = int(input("Choose action: "))

        if choice == 1:
            add()
        elif choice == 2:
            edit()
        elif choice == 3:
            delete()
        elif choice == 4:
            view()
        elif choice == 5:
            return
        else:
            print("Invalid Action.")
```

### `add()` Function Summary

This function allows an instructor to add a new student record. It prompts for the student's name, gender (validated to be 'M' or 'F'), and grades for quiz, recitation, and exam (each validated to be between 0 and 100). It then calculates the `finalGrade` using a weighted average. The new `StudentInfo` object is placed in the first available slot (`occupancy = False`) in the `students` list, assigned a unique ID, and the global `counter` of active students is incremented.

```python
def add():
    global counter
    if counter >= studentSize:
        print("Cannot add more students. Maximum capacity reached.")
        return

    # Find the first available slot
    index = -1
    for i, s in enumerate(students):
        if not s.occupancy:
            index = i
            break

    if index == -1: # Should not happen if counter < studentSize, but good for safety
        print("No available slots found.")
        return

    new_student = students[index]

    new_student.name = input("Enter Student Name: ")

    while True:
        gender_input = input("Enter Student Gender (M/F): ").upper()
        if gender_input in ['M', 'F']:
            new_student.gender = gender_input
            break
        else:
            print("Invalid gender. Please enter 'M' or 'F'.")

    while True:
        try:
            student_id = int(input("Enter Student ID: "))
            # Check for uniqueness
            if any(s.id == student_id and s.occupancy for s in students):
                print("Student ID already exists. Please enter a unique ID.")
            else:
                new_student.id = student_id
                break
        except ValueError:
            print("Invalid ID. Please enter a number.")

    while True:
        try:
            quiz_grade = float(input("Enter Quiz Grade (0-100): "))
            if 0 <= quiz_grade <= 100:
                new_student.quiz = quiz_grade
                break
            else:
                print("Grade must be between 0 and 100.")
        except ValueError:
            print("Invalid input. Please enter a number.")

    while True:
        try:
            recitation_grade = float(input("Enter Recitation Grade (0-100): "))
            if 0 <= recitation_grade <= 100:
                new_student.recitation = recitation_grade
                break
            else:
                print("Grade must be between 0 and 100.")
        except ValueError:
            print("Invalid input. Please enter a number.")

    while True:
        try:
            exam_grade = float(input("Enter Exam Grade (0-100): "))
            if 0 <= exam_grade <= 100:
                new_student.exam = exam_grade
                break
            else:
                print("Grade must be between 0 and 100.")
        except ValueError:
            print("Invalid input. Please enter a number.")

    new_student.finalGrade = (new_student.quiz * 0.40) + (new_student.recitation * 0.10) + (new_student.exam * 0.50)
    new_student.occupancy = True
    counter += 1
    print("Student added successfully!")
```

### `edit()` Function Summary

This function enables an instructor to modify an existing student's grades. It first asks for the student's ID. If found, it displays the current quiz, recitation, and exam grades and allows the instructor to enter new grades or type '-1' to keep the existing one. After updating the relevant grades, it recalculates the `finalGrade` for that student. If the student ID is not found, an appropriate message is displayed.

```python
def edit():
    student_id = int(input("Enter Student ID to edit: "))

    index = next((i for i, s in enumerate(students) if s.id == student_id), -1)

    if index == -1:
        print("Student not found.")
        return

    s = students[index]

    new = float(input(f"Quiz ({s.quiz}) new or -1 to keep: "))
    if new != -1:
        s.quiz = new

    new = float(input(f"Recitation ({s.recitation}) new or -1 to keep: "))
    if new != -1:
        s.recitation = new

    new = float(input(f"Exam ({s.exam}) new or -1 to keep: "))
    if new != -1:
        s.exam = new

    s.finalGrade = s.quiz * 0.40 + s.recitation * 0.10 + s.exam * 0.50

    print("Updated successfully!")
```

### `delete()` Function Summary

This function allows an instructor to delete a student record. It prompts for the student ID to be deleted. If the student is found, it asks for confirmation ('Y/N'). If confirmed, the student's entry in the `students` list is reset to a default `StudentInfo` object (effectively clearing the data and setting `occupancy` to `False`), and the global `counter` is decremented. If the student ID is not found or the deletion is cancelled, it provides appropriate feedback.

```python
def delete():
    global counter

    student_id = int(input("Enter Student ID to delete: "))

    index = next((i for i, s in enumerate(students) if s.id == student_id), -1)

    if index == -1:
        print("Student not found.")
        return

    confirm = input("Are you sure? (Y/N): ").upper()

    if confirm == 'Y':
        students[index] = StudentInfo()
        counter -= 1
        print("Deleted successfully.")
    else:
        print("Cancelled.")
```

### `view()` Function Summary

This function allows an instructor to view student records, sorted either in ascending or descending order by `finalGrade`. It prompts the user to choose the sorting order (1 for Ascending, 2 for Descending). It then filters for occupied student slots, sorts them, and calls the `display()` function to print the formatted list of student data. Invalid choices are handled by displaying an 'Invalid' message.

```python
def view():
    print("(1) View by Names (Alphabetical)")
    print("(2) View by Grades (Highest to Lowest)")

    choice = int(input("Choose: "))

    if choice == 1:
        sorted_list = sorted([s for s in students if s.occupancy], key=lambda x: x.name)
    elif choice == 2:
        sorted_list = sorted([s for s in students if s.occupancy], key=lambda x: x.finalGrade, reverse=True)
    else:
        print("Invalid.")
        return

    display(sorted_list)
```

### `display()` Function Summary

This function is responsible for printing a formatted table of student data to the console. It takes a list of `StudentInfo` objects as input (`data`). It first prints a header row with column names (Name, ID, Quiz, Recitation, Exam, Final) and a separator. Then, it iterates through each student object in the provided list and prints their details, ensuring consistent formatting with tab (`\t`) characters. Finally, it prints another separator line to visually close the table.

```python
def display(data):
    # Adjusted widths for consistent alignment
    print(f"\n{'Name':<15}{'Gender':<8}{'ID':<12}{'Quiz':<10}{'Recitation':<12}{'Exam':<10}{'Final':<10}")
    print("*" * 77)  # Adjusted separator length to match total width

    for s in data:
        # Format grades to two decimal places
        print(f"{s.name:<15}{s.gender:<8}{s.id:<12}{s.quiz:<10.2f}{s.recitation:<12.2f}{s.exam:<10.2f}{s.finalGrade:<10.2f}")

    print("*" * 77)  # Closing separator
```

### `studentMenu()` Function Summary

This function presents a menu to a student after they have successfully logged in using their student ID. It first verifies the provided `student_id`. If valid, it enters a loop offering two options:

*   **(1) View Grades**: Calls the `studentView()` function to display the individual student's grades.
*   **(2) Log-out**: Exits the student menu and returns to the main `userType()` loop.

Invalid choices lead to an 'Invalid' message. If the student ID is not found, an appropriate message is printed and the function returns.

```python
def studentMenu(student_id):
    index = next((i for i, s in enumerate(students) if s.id == student_id), -1)

    if index == -1:
        print("Student not found.")
        return

    while True:
        print("\nStudent Menu")
        print("(1) View Grades")
        print("(2) Log-out")

        choice = int(input("Choose: "))

        if choice == 1:
            studentView(index)
        elif choice == 2:
            return
        else:
            print("Invalid.")
```
