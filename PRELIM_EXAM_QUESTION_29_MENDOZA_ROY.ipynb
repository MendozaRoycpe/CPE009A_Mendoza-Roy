```python
S=[] # array for storing the scores

#--------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------#
n = int(input("HOW MANY SUBJECTS ARE TO BE GRADED?: ")) #asking the user for the amout of grades to be inputted

#--------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------#
print('-' * 20)
for i in range(n): #loop the asking using the "n" for how many iterations of the loop
  score = int(input(f"Enter SUBJECT {i+1}: ")) #
  S.append(score) #storing the grades in "S" array

#--------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------#
print('-' * 20)
print("IS THIS THE FINAL SCORES FOR EACH SUBJECTS?(YES/NO): ")
if input().upper() == "YES": # manual breaking, changed to input of the user to .upper()
  pass
else:
  print('-' * 20)
  print("RE-ENTER THE SCORES: ")

  print('-' * 20)
  S.clear() # clean the array S
  for i in range(n): #loop the asking using the "n" for how many iterations of the loop
    score = int(input(f"Enter SUBJECT {i+1}: "))
    S.append(score) #storing the grades in "S" array again

print(f"THE SCORES ARE: {S}") # printing the content of "S" array

#--------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------#
#code for individual subject grades
print('-' * 20)
print("\n INDIVIDUAL SUBJECT GRADES: ")
for i, score in enumerate(S):
    if score > 100: grade='INVALID'
    elif score > 89.9:grade='A+'
    elif score > 79.9:grade= 'A'
    elif score > 69.9:grade='B+'
    elif score > 59.9:grade= 'B'
    elif score > 49.9:grade='C'
    elif score > 39.9:grade='D'
    else: grade = 'F'
    print(f"SUBJECT {i+1}: Score = {score}, Grade = {grade}")

#--------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------#
# calculating the Average for all the subject
print('-' * 20)
if n > 0:
  print(f"The average score is: {sum(S) / n}") # calculating the average of all subject for the grade remarks
else:
  print("Cannot calculate average: No scores to grade.")

#--------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------#
#Determining the grade wheter passed or not and its Equivalent Grade Marks.
print('-' * 20)
if n > 0:
  if sum(S)/n > 100: print("Invalid Score")
  elif sum(S)/n > 89.9: print(" A+, [You Passed]")
  elif sum(S)/n > 79.9: print(" A, [You Passed]")
  elif sum(S)/n > 69.9: print(" B+, [You Passed]")
  elif sum(S)/n > 59.9: print(" B. [You Passed]")
  elif sum(S)/n > 49.9: print(" C, [You Passed]")
  elif sum(S)/n > 39.9: print(" D, [You Passed]")
  else: print("Failed", " F")
else:
    print("Cannot determine overall grade: No scores to grade.")

#--------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------#
#Determining the Performance Remarks
if n > 0:
  if sum(S)/n > 100: print("Invalid Score")
  elif sum(S)/n > 89.9: print("Outstanding")
  elif sum(S)/n > 79.9: print("Excellent")
  elif sum(S)/n > 69.9: print("Very Good")
  elif sum(S)/n > 59.9: print("Good")
  elif sum(S)/n > 39.9: print("Satisfactory")
  else: print("Needs Improvement")
else:
    print("Cannot determine overall remark: No scores to grade.")
```
