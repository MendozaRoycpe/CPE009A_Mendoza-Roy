**LOOP STATEMENTS IN PYTHON   2-19-26**

LOOP CONTROL STATEMENTS - used to perform looping operations until the given condition is TRUE.


KINDS OF LOOPS:



>  

> For Loop - used for iterating over a sequence (that is either a list, tuple, dictionary, set or string) similar to while statements but this statement is to process list in large range of numbers.

```python
#EX 1:

week = ["Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"]

for x in week:
  print(x)
```

```python
#EX 2: the break statement

week = ["Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"]
for day in week:
  print(day)
  if day == "Thursday":
        break
```

```python
#EXAMPLE 3:

week = ["Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"]
for x in week:
  if x == "Thursday":
    break
print(x) #Check the indention for the print output to ensure whether all the x or just the x
```

```python
#EX 4: looping through a string

for x in "week":
  print(x)
```

**THE RANGE FUNCTION:**

> loop through a set of code a specified no. of times we can use the range function.

> The range() function returns a sequence of no. starting from 0 by default and increments by 1(by default), and ends at a specified no.

> Add blockquote



```python
#EX 1 range loop

for x in range(6):
  print(x)
for x in range(2,6):
  print("\n",x)
```

```python
#EXAMPLE 2:

for x in range(2,11,2):
  print(x)
```

```python
#NESTED LOOPS:

adjective=['red', 'big', 'tasty']
fruits=['apple', 'banana', 'cherry']

for x in adjective:
  for y in fruits:
    print(x,y)
```

PYTHON WHILE LOOPS:

>

```python
#EXAMPLE 5

i=1
while i < 6 :
  print(i)
  i+=1#i=i+1 Assignment operator
```

```python
#The break statement of while loop

i=1
while i<6:
  print(i)
  if i==3:
    break
  i+=1
```

```python
#Continue Statement

i=0
while i<6:
  i+=1
  if i==3:
    print(i)
  continue
```

```python
#The else statement

i=1
while i<6:
  print(i)
  i+=1
else:
  print("i is no longer less than 6")
```
