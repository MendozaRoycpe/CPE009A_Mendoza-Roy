```python
# Distance Formula
#1. Make a program that would calculate the distance between two points given a list of coordinates. Use the distance
#formula.
#coorindates_list = [(1,1), (2,3)]
import math

# Get user input for coordinates first
final_x = int(input("Enter final x: "))
initial_x = int(input("Enter initial x: "))
final_y = int(input("Enter final y: "))
initial_y = int(input("Enter initial y: "))

# where
x2 = final_x
x1 = initial_x
y2 = final_y
y1 = initial_y

coordinateFormula = math.sqrt((x2-x1)**2 + (y2-y1)**2)

print(coordinateFormula)
```

```python
#Simple Word Filter
#2. For a given string input, replace all the words “stupid” with an asterisk * equal to the length of the string. The new
#string value should be displayed with the asterisks.

inval = {'stupid': '******', 'animal':'******' } # the dictionary for what to censor
initial_string = input("Enter a string: ") # asking the user for a string

# Initialize filtered_string with the initial input
filtered_string = initial_string

import re

# Iterate through each word and its replacement in the dictionary
for word_to_filter, replacement_pattern in inval.items():
    # Using re.sub for case-insensitive and whole word matching, I got this online since i do not know how to filter it and there is actually a re library
    filtered_string = re.sub(r'\b' + re.escape(word_to_filter) + r'\b', replacement_pattern, filtered_string, flags=re.IGNORECASE)

print("Filtered string:", filtered_string)
```

```python
#Phonebook
#3. Create a simple phonebook program that can read from a list of dictionaries. The program should be able to display a
#person’s name, contact, and address based from a user input which is the id of the record.

phonebook = [
    {'id': 1, 'name': 'Roy Gabriel Mendoza', 'contact': '09061234567', 'address': 'TIP'},
    {'id': 2, 'name': 'Ryo Shin Mendoza', 'contact': '09061234568', 'address': 'TIP'},
    {'id': 3, 'name': 'Allyson Mendooza', 'contact': '09061234569', 'address': 'TIP'}
] # phonebook dictionary content

def find(id):
  for person in phonebook:
    if person['id'] == id:
      return person
  return None

id = int(input("Enter the ID of the person to search: "))
person = find(id)

print("The person is")
print(person['name'],  person['contact'],  person['address'])

```
