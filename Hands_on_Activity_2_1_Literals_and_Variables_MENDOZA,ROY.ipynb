**Procedure:**

1.   A teacher wants to calculate the final grade in a CpE course and want to write it in a python program. The following are the requirements:

>> PRELIM GRADE = 50% Prelim Exam + 50% Prelim Class Standing (CS)

>>PRELIM CS = 50% Hands-on activity + 30% Quiz + 20% Assignment

>> MIDTERM GRADE = 1/3 of PRELIM GRADE + 2/3 of (50% Midterm Exam + 50% Midterm Class Standing (CS))

>> MIDTERM CS = 50% Hands-on activity + 30% Quiz + 20% Assignment

>> FINAL GRADE = 1/3 of MIDTERM GRADE + 2/3 of (50% Final Exam + 50% Final Class Standing (CS))

>> FINAL CS = 50% Hands-on activity + 30% Quiz + 20% Assignment
HOAs, Quizzes and Assignments are inputted as average of all submissions and are out of 100%.

>> Major exams are inputted out of 100%.

>> Show the codes that successfully run the program.

>> Provide comments or documentation strings for your program.

```python
# Place all the output of the procedures here
#Mendoza, Roy Gabriel I. - CPE12S2
#code


#PRELIM DATA
PreHOA = float(input("Enter PRELIM Hands-on Activity grade: "))
PreQUIZ = float(input("Enter Prelim Quiz grade: "))
PreASSIGN = float(input("Enter Prelim Assignment grade: "))
PreEX = float(input("Enter Prelim Exam grade: "))

#PRELIM computations
PreCS = float(PreHOA*0.5 + PreQUIZ*0.3 + PreASSIGN*0.2) #NEED INPUT OF (HOA><QUIZ><ASSIGNMENT><) for prelims
PreGR = float(PreEX*.5 + PreCS*.5)


#MIDTERM DATA
MidHOA = float(input("Enter MIDTERM Hands-on Activity grade: "))
MidQUIZ = float(input("Enter Midterm Quiz grade: "))
MidASSIGN = float(input("Enter Midterm Assignment grade: "))
MidEX = float(input("Enter Midterm Exam grade: "))

#MID TERM computations
MidCS = float(MidHOA*0.5 + MidQUIZ*0.3 + MidASSIGN*0.2) #NEED INPUT OF (HOA><QUIZ><ASSIGNMENT><) for midterm
MidGR = float(PreGR*0.3 + (MidEX*0.5 + MidCS*0.5)*2/3)


#Finals DATA
FinHOA = float(input("Enter MIDTERM Hands-on Activity grade: "))
FinQUIZ = float(input("Enter Final Quiz grade: "))
FinASSIGN = float(input("Enter Final Assignment grade: "))
FinEX = float(input("Enter Final Exam grade: "))

#FINALS computations
FinCS = float(FinHOA*0.5 + FinQUIZ*0.3 + FinASSIGN*0.2) # #NEED INPUT OF (HOA><QUIZ><ASSIGNMENT><) for finals
FinGR = float(MidGR*0.3+(FinEX*0.5 + FinCS*0.5)*2/3)


#PRINTING OF OUTPUTS
#PRELIMS
print("-----------------------------------------------------------------------------------------------------------")
print("\n The Prelim Grade is: "),print (PreGR),print("\n The Prelim Class Standing is: ") ,print(PreCS)
print("\n The Midterm Grade is: "),print(MidGR), print("\n The Midterm Class Standing is: "),print(MidCS)
print("\n The Final Grade is: "),print(FinGR), print("\n The Final Class Standing is: "),print(FinCS)
```
