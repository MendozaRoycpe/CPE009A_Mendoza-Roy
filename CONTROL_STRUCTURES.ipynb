IF STATEMENT

```python
a=12
b=100
if b>a:
  print("b is greater than a")
```

ELIF STATEMENT

```python
a=12
b=100
if b>a:
  print("b is greater than a")
elif a==b:
  print("a and b are equal")
```

ELSE STATEMENT

```python
a=12
b=100
if b>a:
  print("b is greater than a")
elif a==b:
  print("a and b are equal")
else:
  print("a is greater than b")
```

AND-BOTH CONDITIONS ARE TRUE

```python
a=200
b=33
c=500
if a>b and c>a:
  print("Both conditions are True")
```

OR-ATLEAST ONE OF THE CONDITIONS ARE TRUE

```python
a=200
b=33
c=500
if a>b or a>c:
  print("At least one of the conditions is True")
```
